```

# Import libraries and datasets
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
from statsmodels.stats import power

sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (8,4)

# Load datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')

print('Datasets loaded successfully.')
```

## 1️⃣ One-sample t-test


```
sample = tips['total_bill'].dropna()
mu_0 = 19.0
t_stat, p_val = stats.ttest_1samp(sample, mu_0)
print(f'T-Statistic: {t_stat:.4f}, P-value: {p_val:.4f}')
```

## 2️⃣ Two-sample Independent t-test

```
group1 = tips.loc[tips['smoker']=='Yes', 'total_bill']
group2 = tips.loc[tips['smoker']=='No', 'total_bill']
t_stat, p_val = stats.ttest_ind(group1, group2, equal_var=False)
print(f'T-statistic: {t_stat:.4f}, p-value: {p_val:.4f}')
```

## 3️⃣ Paired t-test (synthetic example)

```
np.random.seed(0)
before = np.random.normal(50,5,30)
after = before + np.random.normal(-1.5,3,30)
t_stat, p_val = stats.ttest_rel(before, after)
print(f'Paired t-test: t={t_stat:.4f}, p={p_val:.4f}')
```

## 4️⃣ Two-proportion z-test (Titanic: Male vs Female Survival)

```
male = titanic[titanic['sex']=='male']['survived'].dropna()
female = titanic[titanic['sex']=='female']['survived'].dropna()
n1, n2 = len(male), len(female)
p1, p2 = male.mean(), female.mean()
p_pool = (p1*n1 + p2*n2)/(n1+n2)
z = (p1-p2)/np.sqrt(p_pool*(1-p_pool)*(1/n1+1/n2))
p_val = 2*(1-stats.norm.cdf(abs(z)))
print(f'z={z:.4f}, p={p_val:.6f}')
```

## 5️⃣ One-way ANOVA (Iris sepal length by species)

```
groups = [g['sepal_length'] for _, g in iris.groupby('species')]
f_stat, p_val = stats.f_oneway(*groups)
print(f'F={f_stat:.4f}, p={p_val:.6f}')
```

## 6️⃣ Chi-square Test for Independence (Titanic: Class vs Survival)

```
cont = pd.crosstab(titanic['pclass'], titanic['survived'])
chi2, p, dof, exp = stats.chi2_contingency(cont)
print(f'Chi2={chi2:.4f}, p={p:.6f}, dof={dof}')
```

## 7️⃣ Non-parametric Tests

```
# Mann–Whitney U test (non-parametric alternative to two-sample t-test)
stat, p = stats.mannwhitneyu(group1, group2, alternative='two-sided')
print(f'Mann–Whitney U: statistic={stat:.4f}, p={p:.4f}')
```

```
# Wilcoxon signed-rank test (non-parametric paired test)
stat, p = stats.wilcoxon(before, after)
print(f'Wilcoxon signed-rank: stat={stat:.4f}, p={p:.4f}')
```

```
# Kruskal–Wallis Test (non-parametric ANOVA alternative)
stat, p = stats.kruskal(*groups)
print(f'Kruskal–Wallis: stat={stat:.4f}, p={p:.6f}')
```

## 8️⃣ Effect Size Calculations

```
# Cohen's d for independent samples
d = (group1.mean()-group2.mean()) / np.sqrt((group1.var()+group2.var())/2)
print(f"Cohen's d = {d:.4f}")
```

```
# Eta-squared for ANOVA
ss_between = sum([len(g)*(g.mean()-iris['sepal_length'].mean())**2 for g in groups])
ss_total = sum((iris['sepal_length']-iris['sepal_length'].mean())**2)
eta_sq = ss_between/ss_total
print(f"Eta-squared = {eta_sq:.4f}")
```

## 9️⃣ Power Analysis

```
analysis = power.TTestIndPower()
result = analysis.solve_power(effect_size=0.5, power=0.8, alpha=0.05)
print(f'Required sample size per group for medium effect (0.5): {result:.2f}')
```

## 🔟 A/B Testing Example (Synthetic Data)
Simulate conversion rates for two website versions (A & B).

```
np.random.seed(1)
A = np.random.binomial(1, 0.12, 500)
B = np.random.binomial(1, 0.15, 500)

p1, p2 = A.mean(), B.mean()
n1, n2 = len(A), len(B)
p_pool = (p1*n1 + p2*n2)/(n1+n2)
z = (p1-p2)/np.sqrt(p_pool*(1-p_pool)*(1/n1+1/n2))
p_val = 2*(1-stats.norm.cdf(abs(z)))
print(f'A/B test: z={z:.4f}, p={p_val:.6f}')
```

## ✅ Summary
- Parametric tests assume normality; non-parametric tests don’t.
- Always check sample size, variance equality, and outliers.
- Use effect size and power analysis for interpretation and planning.
- Hypothesis testing is the foundation for A/B testing and model validation.
