# 📊 Bivariate Analysis on Titanic Dataset
This notebook demonstrates **Bivariate Analysis** — studying the relationship between two variables at a time.
We will explore how numerical and categorical features relate to each other using the Titanic dataset.

```

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
data = sns.load_dataset('titanic')

# Display first few rows
data.head()
```

## 🧩 Step 1: Basic Information

```
data.info()
```

## 🔢 Step 2: Numerical vs Numerical Analysis
Let's explore how two numerical variables relate to each other.

```

# Scatter plot for Age vs Fare
plt.figure(figsize=(7,5))
sns.scatterplot(x='age', y='fare', data=data)
plt.title('Age vs Fare')
plt.show()
```

```

# Correlation between Age and Fare
data[['age', 'fare']].corr()
```

## 🧠 Step 3: Categorical vs Categorical Analysis
Example: Relationship between Sex and Survival.

```

# Cross-tabulation
pd.crosstab(data['sex'], data['survived'])
```

```

# Countplot visualization
sns.countplot(x='sex', hue='survived', data=data)
plt.title('Survival Count by Gender')
plt.show()
```

## ⚙️ Step 4: Numerical vs Categorical Analysis
Example: Relationship between Age and Survival.

```

sns.boxplot(x='survived', y='age', data=data)
plt.title('Age Distribution by Survival')
plt.show()
```

```

sns.violinplot(x='survived', y='fare', data=data)
plt.title('Fare Distribution by Survival')
plt.show()
```

## 🧮 Step 5: Correlation Matrix (For All Numerical Features)

```

corr = data.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 🧩 Step 6: Pairplot (Visualize Multiple Relationships Together)

```

sns.pairplot(data[['age', 'fare', 'survived']], hue='survived')
plt.show()
```

## ✅ Step 7: Key Insights
| Observation | Insight |
|--------------|----------|
| Age vs Fare | Weak correlation between age and fare. |
| Sex vs Survived | Females had higher survival rate. |
| Age vs Survived | Survivors tend to be slightly younger. |
| Fare vs Survived | Passengers who paid higher fares had higher survival chance. |


### 🧭 Summary:
- Bivariate Analysis explores how two variables relate to each other.
- It helps detect correlations, associations, and feature importance.
- Useful for identifying key predictors for machine learning models.
