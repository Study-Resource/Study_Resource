# Sampling Demonstration using Seaborn Diamonds Dataset
**Objective:** Understand and visualize different sampling techniques, their implementation, and comparison with population statistics.
---

```

# Import necessary libraries
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math

# Set display options
pd.set_option('display.float_format', lambda x: '%.2f' % x)
sns.set(style="whitegrid")

# Load the diamonds dataset
diamonds = sns.load_dataset('diamonds')
print(f"Dataset shape: {diamonds.shape}")
diamonds.head()
```

## 1️⃣ Exploratory Data Analysis on Full Population

```

# Check basic stats
diamonds.describe()

# Distribution of 'cut' and 'color'
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(data=diamonds, x='cut', ax=axes[0])
sns.countplot(data=diamonds, x='color', ax=axes[1])
axes[0].set_title('Distribution of Cut')
axes[1].set_title('Distribution of Color')
plt.tight_layout()
plt.show()
```

## 2️⃣ Simple Random Sampling (SRS)

```

# Draw a simple random sample
srs_sample = diamonds.sample(n=1000, random_state=42)

# Compare distributions
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(diamonds['price'], bins=40, color='lightblue', label='Population', ax=ax[0])
sns.histplot(srs_sample['price'], bins=40, color='orange', label='Sample', ax=ax[1])
ax[0].set_title('Population Price Distribution')
ax[1].set_title('SRS Sample Price Distribution')
plt.show()

# Mean comparison
print("Population mean price:", diamonds['price'].mean())
print("Sample mean price:", srs_sample['price'].mean())
```

## 3️⃣ Stratified Sampling by 'cut'

```

# Stratified sample by 'cut'
stratified_sample = diamonds.groupby('cut', group_keys=False).apply(lambda x: x.sample(frac=0.1, random_state=42))

# Compare mean prices per stratum
pop_means = diamonds.groupby('cut')['price'].mean()
sample_means = stratified_sample.groupby('cut')['price'].mean()

compare_df = pd.DataFrame({'Population Mean': pop_means, 'Sample Mean': sample_means})
print(compare_df)

# Visual comparison
compare_df.plot(kind='bar', figsize=(8, 5), title='Stratified Sampling: Mean Price by Cut')
plt.show()
```

## 4️⃣ Systematic Sampling

```

# Determine step size
k = int(len(diamonds) / 1000)
systematic_sample = diamonds.iloc[::k, :]

print(f"Systematic sample size: {len(systematic_sample)}")

# Compare mean prices
print("Population mean price:", diamonds['price'].mean())
print("Systematic sample mean price:", systematic_sample['price'].mean())
```

## 5️⃣ Cluster Sampling by 'color'

```

# Randomly select 3 clusters (colors)
clusters = diamonds['color'].unique()
selected_clusters = np.random.choice(clusters, 3, replace=False)
cluster_sample = diamonds[diamonds['color'].isin(selected_clusters)]

print(f"Selected clusters: {selected_clusters}")
print(f"Cluster sample size: {len(cluster_sample)}")

# Compare distributions
sns.countplot(data=cluster_sample, x='color')
plt.title('Cluster Sampling: Selected Colors')
plt.show()
```

## 6️⃣ Sample Size Calculation Examples

```

# Example: Estimating mean price with 95% confidence, margin of error = 200
z = 1.96
sigma = diamonds['price'].std()
E = 200
n = (z * sigma / E) ** 2
print(f"Required sample size for mean estimation: {math.ceil(n)}")

# Finite population correction
N = len(diamonds)
n_adj = n / (1 + (n - 1) / N)
print(f"Adjusted for finite population: {math.ceil(n_adj)}")
```

## 7️⃣ Summary of Sampling Techniques

```

summary = pd.DataFrame({
    'Method': ['Simple Random', 'Stratified', 'Systematic', 'Cluster'],
    'Sample Size': [len(srs_sample), len(stratified_sample), len(systematic_sample), len(cluster_sample)],
    'Mean Price': [
        srs_sample['price'].mean(),
        stratified_sample['price'].mean(),
        systematic_sample['price'].mean(),
        cluster_sample['price'].mean()
    ]
})

print(summary)
```
