# 🧠 Univariate Analysis on Titanic Dataset
This notebook demonstrates **Univariate Analysis** — analyzing one variable at a time to understand its distribution, spread, and key patterns.

```

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
data = sns.load_dataset('titanic')

# Display first few rows
data.head()
```

## 🔢 Step 2: Identify Variable Types

```
data.info()
```

## 🧮 Step 3: Univariate Analysis – Numerical Variables
### A. Age

```
data['age'].describe()
```

```

plt.figure(figsize=(7,5))
sns.histplot(data['age'], bins=30, kde=True)
plt.title('Age Distribution')
plt.show()
```

```

sns.boxplot(x=data['age'])
plt.title('Boxplot of Age')
plt.show()
```

### B. Fare

```

plt.figure(figsize=(7,5))
sns.histplot(data['fare'], bins=40, kde=True)
plt.title('Fare Distribution')
plt.show()
```

## 🔠 Step 4: Univariate Analysis – Categorical Variables

```

sns.countplot(x='sex', data=data)
plt.title('Count of Passengers by Gender')
plt.show()
```

```

sns.countplot(x='class', data=data)
plt.title('Passenger Count by Class')
plt.show()
```

```

sns.countplot(x='embarked', data=data)
plt.title('Passengers by Embarkation Port')
plt.show()
```

```

sns.countplot(x='survived', data=data)
plt.title('Survival Distribution (0 = No, 1 = Yes)')
plt.show()
```

## 🧩 Step 5: Key Insights
| Observation | Insight |
|--------------|----------|
| Age | Mostly between 20–40 years, few outliers. |
| Fare | Right-skewed; a few paid very high fares. |
| Sex | More males than females. |
| Class | 3rd class most common. |
| Embarked | Most passengers boarded from ‘S’. |
| Survived | Fewer survivors (around 38%). |


✅ **Summary:**
Univariate analysis helps understand:
- The **shape** and **spread** of data.
- The **balance** of categories.
- Potential **missing values**, **outliers**, and **skewness**.

This is an essential foundation for **bivariate** and **multivariate** analysis.
