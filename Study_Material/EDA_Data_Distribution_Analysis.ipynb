# 📊 Understanding Data Distributions in Data Science
This notebook explains **different types of data distributions**, how to identify them, visualize them, and handle them through feature engineering, outlier treatment, and model selection.
---

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set(style='whitegrid')
np.random.seed(42)
```

## 1️⃣ Normal (Gaussian) Distribution

```python

# Generate Normal distribution data
normal_data = np.random.normal(loc=50, scale=10, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(normal_data, kde=True, color='skyblue')
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

print('Mean:', np.mean(normal_data))
print('Skewness:', stats.skew(normal_data))
print('Kurtosis:', stats.kurtosis(normal_data))
```


### 🧠 Insights & Handling
- **Feature Engineering:** Usually no transformation needed; normalization/scaling is fine.
- **Outlier Handling:** Use Z-score or IQR method.
- **Hypothesis Testing:** t-test, ANOVA, Pearson correlation assume normality.
- **Model Suitability:** Linear Regression, Logistic Regression, LDA, Naïve Bayes work well with normally distributed data.


## 2️⃣ Right-Skewed Distribution

```python

right_skewed = np.random.exponential(scale=1.0, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(right_skewed, kde=True, color='orange')
plt.title('Right-Skewed Distribution')
plt.show()

print('Skewness:', stats.skew(right_skewed))
```


### 🧠 Insights & Handling
- **Feature Engineering:** Apply `log`, `sqrt`, or `Box-Cox` transformation to reduce skewness.
- **Outlier Handling:** Use IQR or percentile capping (1% and 99%).
- **Hypothesis Testing:** Use non-parametric tests (Mann–Whitney U, Kruskal–Wallis) if not normal.
- **Model Suitability:** Tree-based models (Decision Trees, Random Forest, XGBoost) handle skewed data well.


## 3️⃣ Left-Skewed Distribution

```python

left_skewed = -np.random.exponential(scale=1.0, size=1000) + 5

plt.figure(figsize=(8,5))
sns.histplot(left_skewed, kde=True, color='green')
plt.title('Left-Skewed Distribution')
plt.show()

print('Skewness:', stats.skew(left_skewed))
```


### 🧠 Insights & Handling
- **Feature Engineering:** Apply power transformation like `x^2` or use `Box-Cox`.
- **Outlier Handling:** Winsorization (clip extreme low values).
- **Hypothesis Testing:** Use transformations before parametric tests.
- **Model Suitability:** Tree-based models again robust; regression may require transformation.


## 4️⃣ Uniform Distribution

```python

uniform_data = np.random.uniform(low=0, high=100, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(uniform_data, kde=True, color='purple')
plt.title('Uniform Distribution')
plt.show()

print('Skewness:', stats.skew(uniform_data))
```


### 🧠 Insights & Handling
- **Feature Engineering:** No transformation needed.
- **Outlier Handling:** Uniform data rarely has outliers.
- **Hypothesis Testing:** Not ideal for parametric tests — use random sampling or segmentation.
- **Model Suitability:** Tree-based and distance-based models (KNN, SVM) work fine.


## 5️⃣ Bimodal Distribution

```python

bimodal = np.concatenate([np.random.normal(40, 5, 500), np.random.normal(70, 5, 500)])

plt.figure(figsize=(8,5))
sns.histplot(bimodal, kde=True, color='teal')
plt.title('Bimodal Distribution')
plt.show()
```


### 🧠 Insights & Handling
- **Feature Engineering:** Split data by group (e.g., gender), or create binary feature if groups identifiable.
- **Outlier Handling:** Analyze each group separately.
- **Hypothesis Testing:** Perform tests within each group separately.
- **Model Suitability:** Clustering, mixture models (GMM), or decision trees handle multimodal data well.


## 6️⃣ Exponential Distribution

```python

exp_data = np.random.exponential(scale=2, size=1000)

plt.figure(figsize=(8,5))
sns.histplot(exp_data, kde=True, color='red')
plt.title('Exponential Distribution')
plt.show()
```


### 🧠 Insights & Handling
- **Feature Engineering:** Apply log transformation for normalization.
- **Outlier Handling:** Cap extreme long tails.
- **Hypothesis Testing:** Use non-parametric tests.
- **Model Suitability:** Tree-based models, Poisson regression for count/time data.


## 🧭 Summary Table

| Distribution | Feature Engineering | Outlier Handling | Hypothesis Tests | Best Models |
|---------------|--------------------|------------------|------------------|--------------|
| Normal | Scaling, standardization | Z-score, IQR | t-test, ANOVA | Linear/Logistic Regression |
| Right-Skewed | Log, sqrt | IQR, capping | Mann–Whitney, Kruskal | Tree models |
| Left-Skewed | Square, Box-Cox | Winsorization | Transformed t-tests | Tree models |
| Uniform | None | Rare | Non-parametric | KNN, SVM |
| Bimodal | Separate groups | Group-wise IQR | Group-wise tests | GMM, Clustering |
| Exponential | Log transform | Capping | Non-parametric | Tree, Poisson Regression |
