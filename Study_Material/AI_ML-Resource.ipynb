## __Statistics and Probability__

1. __List all the tests in hypothesis testing__
<div class="alert alert-block alert-info">
<b>Tip:</b> Reply from Google AI:

Common hypothesis tests include the Z-test, T-test, Chi-Square test, and Analysis of Variance (ANOVA). Other tests include non-parametric options like the Mann-Whitney U test and Kruskal-Wallis test, which are used when data does not meet the assumptions of parametric tests. <br><br>
__Common parametric tests__<br>
__Z-test:__ Used to compare the means of two samples when the population variance is known and sample sizes are large.<br><br>
__T-test:__ Used for smaller sample sizes or when the population standard deviation is unknown. There are one-sample and two-sample versions.<br><br>
__Chi-Square test:__ Used to test the goodness of fit for a distribution or to compare variances. It is also used for analyzing categorical data.<br><br>
__ANOVA:__ Stands for Analysis of Variance. It compares the means of two or more groups to determine if they are statistically different from each other. <br><br>
__Common non-parametric tests__<br>
__Mann-Whitney U test:__ A non-parametric alternative to the two-sample t-test for comparing two independent groups.<br><br>
__Wilcoxon Signed-Rank test:__ A non-parametric alternative to the paired t-test.<br><br>
__Kruskal-Wallis test:__ A non-parametric alternative to a one-way ANOVA for comparing three or more independent groups.<br><br> 
__Other types of tests__<br>
__Normality test:__ Checks if the data is normally distributed, which is an assumption for many parametric tests.<br><br>
__Tests for correlation__: Specifically tests for the correlation between two variables.<br><br>
__Binomial and Poisson tests:__ Tests for data that follows a binomial or Poisson distribution.<br><br>

## __Exploratory Data Analysis__

1. __Is outliers only for numerical columns or can we use it for object data types too?__
<div class="alert alert-block alert-info">
<b>Tip:</b> Reply from Google AI:

Outliers are not exclusive to numerical columns; they can also exist in categorical (object) data types. While the methods for detecting and treating them differ, the concept of an outlier – a data point significantly deviating from the norm – applies to both. <br><br>
__For Numerical Data:__<br>
Outliers are typically identified by their extreme values compared to the rest of the distribution.
Common detection methods include using statistical measures like the Interquartile Range (IQR), Z-scores, or visualizing data with box plots.<br><br>
__For Categorical Data:__<br>
Outliers are often characterized by their extremely low frequency of occurrence within a category or a combination of categories.<br><br>
__Detection methods may involve:__<br>
Frequency Analysis: Identifying categories or combinations of categories that appear very rarely.
Association Rule Mining: Discovering unusual relationships or patterns between categorical variables.
Categorical Outlier Detection Algorithms: Specialized algorithms like STIRR or those based on k-prototypes can be used to identify outliers in datasets with mixed data types or purely categorical data.
Domain Knowledge: Expert knowledge about the data and its expected patterns can be crucial in identifying unusual or erroneous categorical values.<br><br>
__Example:__<br>
Consider a dataset of customer demographics with a "Gender" column. If 99% of customers are marked "Male" or "Female," but a few entries are "Other" or even misspelled variations like "Femal," these infrequent or erroneous entries could be considered categorical outliers.<br>
</div>
