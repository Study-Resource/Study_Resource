# 🌐 Multivariate Analysis on Titanic Dataset
This notebook demonstrates **Multivariate Analysis** — analyzing the relationship among three or more variables simultaneously.
We'll use the Titanic dataset to visualize and interpret multiple feature relationships.

```

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
data = sns.load_dataset('titanic')

# Display first few rows
data.head()
```

## 🧩 Step 1: Basic Overview

```
data.info()
```

## 📈 Step 2: Pairplot (Visualizing Multiple Relationships Together)
This helps us see how multiple numeric variables interact and how they differ by survival status.

```

sns.pairplot(data[['age', 'fare', 'pclass', 'survived']], hue='survived')
plt.show()
```

## 🧮 Step 3: Correlation Heatmap
Shows how all numeric features are correlated with each other.

```

corr = data.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap - Titanic Dataset')
plt.show()
```

## 🧠 Step 4: Multivariate Relationship using Categorical + Numerical + Target Variable
Example: How **Age**, **Fare**, and **Survival** vary across **Passenger Class**.

```

plt.figure(figsize=(8,6))
sns.scatterplot(x='age', y='fare', hue='survived', style='class', data=data)
plt.title('Age vs Fare by Survival and Class')
plt.show()
```

## ⚙️ Step 5: FacetGrid (Visualize Multi-dimensional Data)
FacetGrid allows plotting of multiple subplots based on categorical variables.

```

g = sns.FacetGrid(data, col='sex', row='pclass', hue='survived', height=3)
g.map(sns.scatterplot, 'age', 'fare', alpha=0.7)
g.add_legend()
plt.show()
```

## 🔢 Step 6: Grouped Analysis using Pivot Table
We can summarize data across multiple categorical variables.

```

pd.pivot_table(data, values='survived', index=['sex', 'pclass'], columns='embarked', aggfunc='mean')
```

## 🧭 Step 7: Multivariate Visualization using 3D Plot (Optional)

```

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(data['age'], data['fare'], data['pclass'], c=data['survived'], cmap='coolwarm')
ax.set_xlabel('Age')
ax.set_ylabel('Fare')
ax.set_zlabel('Pclass')
plt.title('3D Plot: Age, Fare, and Class by Survival')
plt.show()
```

## ✅ Step 8: Key Insights
| Observation | Insight |
|--------------|----------|
| Higher Fare + 1st Class + Female | Highest chance of survival. |
| Lower Class + Male | Lowest chance of survival. |
| Younger Passengers | Slightly higher survival tendency. |
| Correlation Matrix | Fare and Pclass are negatively correlated (lower class → lower fare). |


### 🧭 Summary:
- **Multivariate Analysis** examines how three or more features interact.
- It helps detect **hidden patterns** that may not be visible in bivariate analysis.
- It is crucial for **feature selection** and **understanding complex data structures** before building ML models.
